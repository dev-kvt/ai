```
Data Points
      │
      ▼
Guess a Line (w, b)
      │
      ▼
Predict Values
      │
      ▼
Measure Error (MSE)
      │
      ▼
Compute Gradients
      │
      ▼
Update w and b
      │
      ▼
Repeat until Error is Small
      │
      ▼
Use the Final Line for Predictions
```

````python
import numpy as np 

# data 
# Y -> labelled answer 


X = np.array([23, 32, 45, 67, 7777, 888, 999])
Y = np.array([45455, 655, 6777, 88999, 12345, 6789, 9876])  # 7 target values matching 7 inputs

# parameters

weight = 0 
bias = 0 

n = len(X)
learning_rate =0.0000001
# a foundational hyperparameter that dictates how much a machine learning model adjusts its parameters (weights) in response to its calculated error

epochs = 1000
# a hyperparameter in machine learning that defines one complete pass of the entire training dataset through the learning algorithm
for epoch in range(epochs):

    predictions = weight*X + bias



    error = predictions  - Y
    loss = np.mean(error*2)


    # these are gradients
    dw = (2/n) * np.sum(error*X)
    db = (2/n) * np.sum(error)

    weight -= learning_rate*dw 
    bias -= learning_rate*db

    if epoch % 100 == 0:
        print(f"epoch: {epoch}, loss: {loss}")
    
print("Final weight:", weight)
print("Final bias:", bias)
````
