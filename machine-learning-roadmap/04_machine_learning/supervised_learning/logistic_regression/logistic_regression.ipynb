| Linear Regression   | Logistic Regression       |
| ------------------- | ------------------------- |
| Predicts number     | Predicts probability      |
| Output: any value   | Output: 0–1               |
| Straight line       | Sigmoid curve             |
| MSE loss            | Binary Cross Entropy loss |
| Used for regression | Used for classification   |


```python
import numpy as np

# Dataset
X = np.array([1, 2, 3, 4, 5], dtype=float)
Y = np.array([0, 0, 0, 1, 1], dtype=float)

w = 0.0
b = 0.0

learning_rate = 0.1
epochs = 1000



def sigmoid(z):
    return 1 / (1 + np.exp(-z))


n = len(X)

for epoch in range(epochs):

    z = w * X + b
    p = sigmoid(z)

    loss = -np.mean(
        Y * np.log(p + 1e-9) +
        (1 - Y) * np.log(1 - p + 1e-9)
    )

    dw = (1 / n) * np.sum((p - Y) * X)
    db = (1 / n) * np.sum(p - Y)

    w -= learning_rate * dw
    b -= learning_rate * db

print("Weight:", w)
print("Bias:", b)
```
